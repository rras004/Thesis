# I. Different opinion models
In this code we show the figures and results for when the opinion model is changed. This means that we cerate a recursive ER random hypergraph, and run the opinion models with fixed params on it. 

```python
from analysis1 import compare_models_ratios, plot_model_comparison_majority_ratio, plot_all_runs_majority_ratio, plot_model_comparison_inactive_edge_ratio, plot_model_comparison_inactive_edge_ratio, compute_mean_absolute_deviation, print_final_state_summary
```

```python
n= 100 # Number of nodes
K=100 # The number of simulations that we run for each point for each model.
steps = 3000 # Number discrete time steps
prob_positive_opinion= 0.5 # Probability with what a node is of opinion +1 
recursive_ER_kwargs = {"n": n,"p1": 0.15,"p2": 0.3} # Probabilities for the recursive Erdős-Rényi random hypergraph model
noisy_probabilities = {"p": 0.5, "q": 0.5} # Probabilities for noisy voter model
adaptive_probabilities = {"p": 0.5, "q": 0.5, "p_triangle": 0.3} # Probabilities for adaptive voter model
```

```python
results = compare_models_ratios(
    recursive_generator_kwargs=recursive_ER_kwargs,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    max_steps=steps,
    K=K,
    standard_kwargs={},
    noisy_kwargs=noisy_probabilities,
    adaptive_kwargs=adaptive_probabilities,
    base_seed=123
)
```

```python
plot_model_comparison_majority_ratio(results)
plot_all_runs_majority_ratio(results)
plot_model_comparison_inactive_edge_ratio(results)
print_final_state_summary(results)
print(compute_mean_absolute_deviation(results, "majority_ratio"))
```

```python
results_4000 = compare_models_ratios(
    recursive_generator_kwargs=recursive_ER_kwargs,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    max_steps=5000,
    K=K,
    standard_kwargs={},
    noisy_kwargs=noisy_probabilities,
    adaptive_kwargs=adaptive_probabilities,
    base_seed=123
)
```

```python
plot_model_comparison_majority_ratio(results_4000)
plot_all_runs_majority_ratio(results_4000)
plot_model_comparison_inactive_edge_ratio(results_4000)
print_final_state_summary(results_4000)
print(compute_mean_absolute_deviation(results_4000, "majority_ratio"))
```
