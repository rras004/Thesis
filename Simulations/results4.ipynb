# IV. Varying the values of the graph generating models
The opinion spread model is fixed, only the parameters of the random graph generating models are changed.

```python
import numpy as np

from analysis4 import estimate_average_majority_ratio_er_hypergraph_noisy, estimate_average_majority_ratio_recursive_er_adaptive, plot_average_majority_ratio_vs_parameter, estimate_average_majority_ratio_hsbm_noisy_p3
```

```python
# Common values
p_values = np.linspace(0, 1, 20)

N = 5 # simulations are reapeted N times
n= 100 # number of nodes
steps = 500 #number discrete time steps
prob_positive_opinion= 0.5 # Probability with what a node is of opinion +1 
noisy_probabilities = {"p": 0.5, "q": 0.5} # Probabilities for noisy voter model
adaptive_probabilities = {"p": 0.5, "q": 0.5, "p_triangle": 0.3} # Probabilities for adaptive voter model

K=10
```

```python
results_noisy_er_hypergraph_p2 = estimate_average_majority_ratio_er_hypergraph_noisy(
    n=n,
    fixed_probabilities={2: 0.15},
    varying_edge_size=2,
    varying_values=p_values,
    noisy_kwargs=noisy_probabilities,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    num_runs=K,
    max_steps=steps,
    base_seed=123
)

plot_average_majority_ratio_vs_parameter(
    results_noisy_er_hypergraph_p2,
    varying_param_name="2-edge probability",
    title="Noisy voter model on ER random hypergraphs"
)
```

```python
results_noisy_er_hypergraph_p3 = estimate_average_majority_ratio_er_hypergraph_noisy(
    n=n,
    fixed_probabilities={2:0.15, 3: 0.001},
    varying_edge_size=3,
    varying_values=p_values,
    noisy_kwargs=noisy_probabilities,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    num_runs=K,
    max_steps=steps,
    base_seed=123
)

plot_average_majority_ratio_vs_parameter(
    results_noisy_er_hypergraph_p3,
    varying_param_name="3-edge probability",
    title="Noisy voter model on ER random hypergraphs"
)
```

```python
results_adaptive_recursive_p1 = estimate_average_majority_ratio_recursive_er_adaptive(
    n=n,
    fixed_generator_kwargs={"p1": 0.15, "p2": 0.3},
    varying_generator_param="p1",
    varying_values=p_values,
    adaptive_kwargs=adaptive_probabilities,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    num_runs=K,
    max_steps=steps,
    base_seed=123
)

plot_average_majority_ratio_vs_parameter(
    results_adaptive_recursive_p1,
    varying_param_name="p1",
    title="Adaptive voter model on recursive ER simplicial complexes"
)
```

```python
results_adaptive_recursive_p2 = estimate_average_majority_ratio_recursive_er_adaptive(
    n=n,
    fixed_generator_kwargs={"p1": 0.15, "p2": 0.3},
    varying_generator_param="p2",
    varying_values=p_values,
    adaptive_kwargs=adaptive_probabilities,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    num_runs=K,
    max_steps=steps,
    base_seed=123
)

plot_average_majority_ratio_vs_parameter(
    results_adaptive_recursive_p2,
    varying_param_name="p2",
    title="Adaptive voter model on recursive ER simplicial complexes"
)
```

```python
results_hsbm_p3 = estimate_average_majority_ratio_hsbm_noisy_p3(
    n=n,
    fixed_probabilities={2: (0.25, 0.05), 3: (0.0025, 0.0005)},
    p3_values=p_values,
    noisy_kwargs=noisy_probabilities,
    opinion_kwargs={"prob_positive": prob_positive_opinion},
    num_runs=K,
    max_steps=steps
)

plot_average_majority_ratio_vs_parameter(
    results_hsbm_p3,
    varying_param_name="p_3",
    title="Noisy voter model on HSBM: varying p_3"
)
```
